## 1. Palindrome Checker (Strings & Two-Pointer Technique)
Checks if a given string reads the same backward as forward, ignoring casing.
```java
public class PalindromeChecker {
    public static boolean isPalindrome(String str) {
        String clean = str.replaceAll("[^a-zA-Z0-9]", "").toLowerCase();
        int left = 0, right = clean.length() - 1;
        while (left < right) {
            if (clean.charAt(left) != clean.charAt(right)) {
                return false;
            }
            left++;
            right--;
        }
        return true;
    }
    public static void main(String[] args) {
        String test = "A man, a plan, a canal: Panama";
        System.out.println("Is palindrome: " + isPalindrome(test));
    }
}
```

```java
import java.util.HashMap;
import java.util.Map;

public class Fibonacci {
    private static Map<Integer, Long> memo = new HashMap<>();

    public static long getFibonacci(int n) {
        if (n <= 1) return n;
        if (memo.containsKey(n)) return memo.get(n);

        memo.put(n, value);
```

## 3. Caesar Cipher (Basic Encryption & Decryption)
Encrypts and decrypts text by shifting characters along the alphabet.
```java
public class CaesarCipher {
    public static String encrypt(String text, int shift) {
        StringBuilder result = new StringBuilder();
        for (char character : text.toCharArray()) {
            if (Character.isLetter(character)) {
                char base = Character.isUpperCase(character) ? 'A' : 'a';
                result.append((char) ((character - base + shift) % 26 + base));
            } else {
                result.append(character);
            }
        }
        return result.toString();
    }
    public static void main(String[] args) {
        String original = "Hello World!";
        String encrypted = encrypt(original, 4);
        System.out.println("Encrypted: " + encrypted);
    }
}
```

```java
import java.util.Arrays;

public class BubbleSort {
    public static void sort(int[] arr) {
        int n = arr.length;
        for (int i = 0; i < n - 1; i++) {
            for (int j = 0; j < n - i - 1; j++) {
                if (arr[j] > arr[j + 1]) {
                    int temp = arr[j];
                    arr[j] = arr[j + 1];
                    arr[j + 1] = temp;
                }
            }
        }
    }
    public static void main(String[] args) {
        int[] data = {64, 34, 25, 12, 22, 11, 90};
        sort(data);
        System.out.println("Sorted array: " + Arrays.toString(data));
    }
}
```

# 🌐 Computer Networks Study Notes (5th Semester - 2022 Scheme)

This reference guide covers the core theoretical and practical modules of Computer Networks (CN) typically taught in the 5th semester of Computer Science & Engineering curriculum.

---

## Module 1: Introduction & Physical Layer

### 1. Network Architectures & Topologies
*   **OSI Reference Model (7 Layers):** Physical, Data Link, Network, Transport, Session, Presentation, Application.
*   **TCP/IP Suite (4/5 Layers):** Network Access (Link), Internet (Network), Transport, Application.
*   **Topologies:** Mesh, Star, Bus, Ring, and Hybrid.

### 2. Physical Layer Basics
*   **Transmission Media:** Guided (Twisted Pair, Coaxial, Fiber Optic) vs. Unguided (Radio, Microwave, Infrared).
*   **Multiplexing:** FDM (Frequency Division), TDM (Time Division), and WDM (Wavelength Division).

---

## Module 2: Data Link Layer (DLL)

DLL is responsible for node-to-node delivery, framing, error control, and flow control.

### 1. Framing Methods
*   Character Count, Flag bytes with Character/Bit Stuffing.

### 2. Error Detection & Correction
*   **Parity Check**
*   **Checksum**
*   **Cyclic Redundancy Check (CRC):** Based on binary division using a generator polynomial.

### 3. Sliding Window Protocols
*   **Stop-and-Wait ARQ**
*   **Go-Back-N (GBN) ARQ:** Sender can send multiple frames up to window size $W$, but must retransmit all unacknowledged frames if one times out.
*   **Selective Repeat (SR) ARQ:** Only the damaged or lost frame is retransmitted.

---

## Module 3: Network Layer

Responsible for host-to-host delivery and routing packets across multiple networks.

### 1. IP Addressing
*   **IPv4:** 32-bit address. Classful (Classes A, B, C, D, E) vs. Classless Inter-Domain Routing (CIDR).
*   **IPv6:** 128-bit address designed to replace IPv4 due to address exhaustion.
*   **Subnetting:** Dividing a single physical network into multiple logical sub-networks.

### 2. Routing Algorithms
*   **Distance Vector Routing (DVR):** Based on the Bellman-Ford algorithm (suffers from the Count-to-Infinity problem).
*   **Link State Routing (LSR):** Based on Dijkstra's Shortest Path algorithm.

---

## Module 4: Transport Layer

Responsible for process-to-process delivery, flow control, and congestion control.

### 1. Protocols
*   **UDP (User Datagram Protocol):** Connectionless, unreliable, lightweight, no flow or congestion control.
*   **TCP (Transmission Control Protocol):** Connection-oriented, reliable, byte-stream service, manages flow and congestion.

### 2. TCP 3-Way Handshake (Connection Establishment)
1.  **SYN:** Client sends a synchronization segment with a random sequence number $x$.
2.  **SYN-ACK:** Server acknowledges with $ACK = x + 1$ and sends its own sequence number $y$.
3.  **ACK:** Client responds with $ACK = y + 1$.

### 3. Congestion Control Algorithms
*   Slow Start, Congestion Avoidance, Fast Retransmit, and Fast Recovery.

---

## Module 5: Application Layer

Defines how user applications interact with the network services.

*   **DNS (Domain Name System):** Resolves human-readable domain names (e.g., `google.com`) to IP addresses.
*   **HTTP / HTTPS:** Hypertext Transfer Protocol (Port 80/443).
*   **SMTP, POP3, IMAP:** Email transmission and retrieval protocols.

````python
import urllib.request
import xml.etree.ElementTree as ET

def fetch_latest_news():
    url = 'https://feeds.bbci.co.uk/news/world/rss.xml'
    req = urllib.request.Request(
        url,
        headers={'User-Agent': 'Mozilla/5.0'}
    )
    try:
        with urllib.request.urlopen(req) as response:
            data = response.read()

        root = ET.fromstring(data)
        print("📰 LATEST WORLD NEWS HEADLINES:\n" + "="*40 + "\n")

        for i, item in enumerate(root.findall('.//item')[:10], 1):
            title = item.find('title').text
            description = item.find('description').text
            link = item.find('link').text

            print(f"{i}. {title}")
            print(f"   {description}")
            print(f"   Link: {link}\n")

    except Exception as e:
        print(f"Could not fetch the news: {e}")

fetch_latest_news()
````
